```python
def is_prime(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    
    # Check odd divisors up to sqrt(n)
    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2
    
    return True
```

```python
def reverse_string(s):
    return s[::-1]
```

```python
def is_palindrome(s):
    return s == reverse_string(s)
```
