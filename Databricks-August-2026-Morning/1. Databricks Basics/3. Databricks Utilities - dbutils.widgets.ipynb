***text* widget**

```python
dbutils.widgets.help("text")
```

```python
dbutils.widgets.text("fname", "Kanakaraju", "First Name")
```

```python
fname = dbutils.widgets.get("fname")
print(fname)
```

***dropdown* widget**

```python
dbutils.widgets.dropdown(
    "city",
    "Hyderabad", 
    ["Hyderabad", "Bangalore", "Chennai", "Mumbai", "Pune"],
    "City"
)
```

```python
city = dbutils.widgets.get("city")
print(city)
```

***multiselect* widget**

```python
dbutils.widgets.multiselect(
    "languages",
    "Python",
    ["Python", "SQL", "Scala", "R", "C#", "Java"],
    "Progrmming Languages"
)
```

```python
# dbutils.widgets.remove("languages")
```

```python
languages = dbutils.widgets.get("languages")
print(languages)
```

```python
output = fname + " lives in " + city + " and knows " + languages
print(output)
```

```python
dbutils.notebook.exit(output)
```
